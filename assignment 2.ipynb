## Strings Longer Than K

Returns strings with a length greater than k.

```python
def longer_than_k(words, k):
    result = []
    for word in words:
        if len(word) > k:
            result.append(word)
    return result
words = input("please enter a words").split()
k = int(input("please enter k"))
answer = longer_than_k(words, k)
print(answer)



```

## Calculate Area

Calculates the area of a circle or rectangle.

```python
import math
def area(shape, radius, length, width):
    if shape == "circle":
        return math.pi * radius**2
    elif shape == "rectangle":
        return length * width
circle_area = area("circle", 5 , 0 , 0 )
print(circle_area)
rectangle_area = area("rectangle", 0 , 4 , 6 )
print(rectangle_area)


    
```

## Sum Numbers in a List

Adds numeric items and ignores non-numeric values, including booleans.

```python
items = [30 ,True ,15 ,20 ,False]
total = 0
for item in items:
    if  isinstance(item, (int,float)) and not isinstance(item, bool):
        total += item
print("total sum:", total)






```

## Restaurant Class
This program uses a Restaurant class to store menu items, table reservations, and customer orders.

```python
class Restaurant:
    def __init__(self):
        self.menu_items = {}
        self.book_table = []
        self.customer_orders = []
    def add_item_to_menu (self,item_name,price):
        self.menu_items[item_name] = price
    def book_tables (self,customer_name,table_number):
        self.book_table.append({"customer_name":customer_name ,"table_number":table_number})
    def customer_order (self,customer_name,item_name):
        self.customer_orders.append({"customer_name":customer_name,"item_name":item_name})
my_restaurant = Restaurant()
my_restaurant.add_item_to_menu("Pizza", 100)
my_restaurant.add_item_to_menu("Pasta", 120)
my_restaurant.add_item_to_menu("Burger", 150)
my_restaurant.book_tables("ali", 5)
my_restaurant.book_tables("ahmed", 3)
my_restaurant.book_tables("mohamed", 4)
my_restaurant.customer_order("ahmed", "Pizza")
my_restaurant.customer_order("ali", "Burger")
my_restaurant.customer_order("mohamed", "Pasta")
print(my_restaurant.menu_items)
print(my_restaurant.book_table)
print(my_restaurant.customer_orders)







        


          

    






```

## Bike Rental System
This program models a bike rental shop using three classes: customer, Bike, and MainBikeRental.

```python
class customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_name(self):
        return self.name

    def set_name(self, new_name):
        self.name = new_name

    def get_age(self):
        return self.age

    def set_age(self, new_age):
        self.age = new_age


class MainBikeRental:
    def __init__(self):
        self.bikes = []
        self.rentals = []

    def get_bikes(self):
        return self.bikes

    def set_bikes(self, new_bikes):
        self.bikes = new_bikes

    def get_rentals(self):
        return self.rentals

    def set_rentals(self, new_rentals):
        self.rentals = new_rentals

    def add_bike(self, bike):
        self.bikes.append(bike)

    def requestBike(self, customer, number_of_bikes):
        if customer.get_age() <= 6:
            print("Customer must be older than 6 to rent a bike.")
            return None
        elif number_of_bikes <= 0:
            print("Number of bikes must be greater than 0.")
            return None
        elif number_of_bikes > len(self.bikes):
            print("Wait for 10 minutes.")
            return None
        else:
            rented_bikes = self.bikes[:number_of_bikes]
            self.bikes = self.bikes[number_of_bikes:]

            self.rentals.append({
                "customer_name": customer.get_name(),
                "number_of_bikes": number_of_bikes,
                "bikes": rented_bikes
            })

            print("Rental successful.")
            return rented_bikes

    def totalCost(self):
        total_cost = 0
        for bike in self.bikes:
            total_cost += bike.get_price()
        return total_cost

    def returnBike(self, rental_time, returned_bikes):
        total_bill = rental_time * 40
        self.bikes.extend(returned_bikes)
        return total_bill


class Bike:
    def __init__(self, bike_name, price):
        self.bike_name = bike_name
        self.price = price

    def get_bike_name(self):
        return self.bike_name

    def set_bike_name(self, new_bike_name):
        self.bike_name = new_bike_name

    def get_price(self):
        return self.price

    def set_price(self, new_price):
        self.price = new_price


shop = MainBikeRental()
bike1 = Bike("City Bike", 500)
shop.add_bike(bike1)

customer1 = customer("Ali", 20)
rented_bikes = shop.requestBike(customer1, 1)

if rented_bikes is not None:
    bill = shop.returnBike(3, rented_bikes)
    print("Bill:", bill)
    print("Returned bike:", rented_bikes[0].get_bike_name())

print("Total cost:", shop.totalCost())
```
