## Normalize a 3×3 NumPy Array

Objective: Create a 3×3 array with random integer values, find the minimum and maximum values, then normalize the array.

```python
import numpy as np 
arr = np.random.randint(1,101, size=(3,3))
min_val = arr.min()
max_val = arr.max()
print("minimum:", min_val)
print("maximum:", max_val)
normalized_arr = (arr - min_val) / (max_val - min_val)
print("normalized array")
print(normalized_arr)
```

## Find the Closest Value in a NumPy Array

Objective: Create a 1D array of 10 random integers, ask the user to enter a number, and find the closest value in the array.

```python
import numpy as np
arr = np.random.randint(1,21, size=(10))
x = int(input("please enter a number"))
dif = np.abs(arr - x)
index = dif.argmin()
closest_value = arr[index]
print("Array", arr)
print("closest value:",closest_value)
```

## Find the Second Largest Element in Each Column

Objective: Create a 5×5 NumPy array with random integers and find the second largest value in each column.

```python
import numpy as np
arr = np.random.randint(1,10,size=(5,5))
sorted_arr = np.sort(arr,axis=0)
second_largest = sorted_arr[-2]
print("array:",arr)
print("sorted arr:",sorted_arr)
print("Result:",second_largest)
```

## Find Rows with Even Sums in a NumPy Array

Objective: Create a 6×4 NumPy array, calculate the sum of each row, and find the rows whose sums are even.

```python
import numpy as np
arr = np.random.randint(1,10,size=(6,4))
rows_sums = np.sum(arr,axis=1)
even_rows = np.where(rows_sums %2==0)
print("Array:",arr)
print("rows sums:",rows_sums)
print("even rows:",even_rows)
```

## Replace Multiples of 3 or 5 in a NumPy Array

Objective: Create a 1D NumPy array containing 30 random integers, then replace all numbers divisible by 3 or 5 with -1.

```python
import numpy as np
arr = np.random.randint(1,51,size =(30))
arr[(arr %3==0) | (arr%5==0) ]= -1
print(arr)

```

## Subtract the Mean of Each Row in a NumPy Array

Objective: Create a 3×3 NumPy array, calculate the mean of each row, and subtract that mean from every element in its corresponding row.

```python
import numpy as np
arr = np.random.randint(1,20,size=(3,3))
rows_average = np.average(arr,axis=1)
rows_average_reshape = rows_average.reshape(3,1)
result = arr - rows_average_reshape
print("Array:",arr)
print("rows average:",rows_average)
print("rows average reshape:",rows_average_reshape)
print("result:",result)
```
