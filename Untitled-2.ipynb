```python
import numpy as np 
arr = np.random.randint(1,101, size=(3,3))
min_val = arr.min()
max_val = arr.max()
print("minimum:", min_val)
print("maximum:", max_val)
normalized_arr = (arr - min_val) / (max_val - min_val)
print("normalized array")
print(normalized_arr)
```

```python
import numpy as np
arr = np.random.randint(1,21, size=(10))
x = int(input("please enter a number"))
dif = np.abs(arr - x)
index = dif.argmin()
closest_value = arr[index]
print("Array", arr)
print("closest value:",closest_value)
```

```python
import numpy as np
arr = np.random.randint(1,10,size=(5,5))
sorted_arr = np.sort(arr,axis=0)
second_largest = sorted_arr[-2]
print("array:",arr)
print("sorted arr:",sorted_arr)
print("Result:",second_largest)
```

```python
import numpy as np
arr = np.random.randint(1,10,size=(6,4))
rows_sums = np.sum(arr,axis=1)
even_rows = np.where(rows_sums %2==0)
print("Array:",arr)
print("rows sums:",rows_sums)
print("even rows:",even_rows)
```
